**Operator Matematika dalam Python**

Operator Aritmatika

```python
x = 8
y = 2

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x // y)
print(x % y)
print(x ** y)
```

Operator Perbandingan

```python
x = 12
y = 2

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

3. Membuat persamaan matematika dan menghitung total uang yang dibutuhkan

```python
buku = 3000
pensil = 1000
penghapus = 500

Agung = (3 * buku) + (2 * pensil)
Sandy = (2 * buku) + (4 * penghapus)
total = Agung + Sandy

print(Agung)
print(Sandy)
print(total)
```
