1. Membuat program untuk menentukan nilai maksimum dari tiga buah bilangan.

```python
x = 3
y = 8
z = 10

nilai_maksimum = max(x, y, z)

print(nilai_maksimum)
```

2. Membuat kode program untuk kasus flowchart

```python
# 1. Meminta input dari pengguna
bil = int(input("Masukkan Bilangan: "))

# 2. Struktur Percabangan sesuai Flowchart
if bil > 0:
    print("Positif")
elif bil < 0:
    print("Negatif")
elif bil == 0:
    print("Nol")
```
